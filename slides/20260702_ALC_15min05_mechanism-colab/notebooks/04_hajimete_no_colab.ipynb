## 1. Colab ってなに？

Google Colab（コラボ）は、**ブラウザの中で Python を書いて実行できる、無料のクラウド環境**です。

- 自分の PC に **何もインストールしません**（環境構築が不要）
- ネットの先にある **Google のコンピュータを借りて** 計算します
- だから、手元の PC が非力でも **重い計算（機械学習など）ができる**
- 機械学習に使う **GPU も無料** で使えます

ノートブックは **コード＋説明文＋結果** を 1 つにまとめられ、**Google ドライブに保存**して、
Google ドキュメントのように **リンクで共有** できます。

## 3. 変数と計算（Python の超基本）

`=` で値に名前をつけられます（これを **変数** といいます）。
`#` から後ろは **コメント**（実行されないメモ）です。

```python
# 変数に値を入れて計算する
students = 40        # 学生の人数
groups   = 5         # 班の数
per_group = students / groups
print('1班あたり', per_group, '人')

# リスト（値の並び）と for ループ
scores = [82, 75, 91, 68, 88]
print('平均点:', sum(scores) / len(scores))
for i, s in enumerate(scores, 1):
    print(f'  {i}人目: {s}点')
```

```python
# 小さな表（データフレーム）を作って表示する
df = pd.DataFrame({
    '科目': ['国語', '数学', '英語', '理科', '社会'],
    '平均点': [72, 65, 80, 68, 74],
})
df  # 最後に変数名を書くと、きれいな表で表示される
```

## 5. GPU を使う（重い計算を速く）

機械学習などの重い計算は **GPU** を使うと速くなります。切り替え方：

1. 上のメニュー **`ランタイム` → `ランタイムのタイプを変更`**
2. **ハードウェア アクセラレータ** で **GPU**（例：T4）を選んで **保存**
3. 右上の **「接続」** で接続

下のセルで、いま GPU が使えるか確認できます（GPU でなくてもこのノートは動きます）。

```python
# いま GPU が割り当てられているか確認（GPUでなくてもこのノートは動きます）
import subprocess
try:
    r = subprocess.run(['nvidia-smi', '--query-gpu=name,memory.total',
                        '--format=csv,noheader'], capture_output=True, text=True)
    if r.returncode == 0 and r.stdout.strip():
        print('✅ GPU が使えます:', r.stdout.strip())
    else:
        print('ℹ️ いまは CPU です（GPUにするには ランタイム→タイプを変更→GPU）')
except FileNotFoundError:
    print('ℹ️ いまは CPU です（GPUにするには ランタイム→ランタイムのタイプを変更→GPU）')
```

## 7. 保存と共有 ／ 困ったとき

### 残す・共有する
- **保存**：メニュー `ファイル → ドライブにコピーを保存`（自分の Google ドライブに残る）
- **共有**：右上の **共有** ボタンからリンクで共有（コメント・共同編集も）

### 困ったときの早見表
- **赤いエラーが出た** → エラーの **最後の行** を読む → セルを **上から実行し直す**
- **全部おかしくなった** → メニュー `ランタイム → ランタイムを再起動` → もう一度上から ▶
- **遅い・止まる** → GPU になっているか／接続が切れていないか
- **コードの意味が分からない** → そのコードを選んで、Colab の **Gemini に質問**
